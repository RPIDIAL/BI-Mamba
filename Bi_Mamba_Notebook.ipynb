```python
import os
import json
import numpy as np
from PIL import Image
from tqdm import tqdm
from medmnist import ChestMNIST

# 1. Configuration des chemins
OUTPUT_DIR = '/content/chest_images'
JSON_PATH = '/content/data/nlst_splits.json'
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(os.path.dirname(JSON_PATH), exist_ok=True)

# 2. Charger ChestMNIST (Version 224px pour votre modèle Vim)
print("Downloading ChestMNIST (224px)...")
train_dataset = ChestMNIST(split="train", download=True, size=224)
val_dataset = ChestMNIST(split="val", download=True, size=224)
test_dataset = ChestMNIST(split="test", download=True, size=224)

def save_and_create_entries(dataset, split_name):
    entries = []
    print(f"Processing {split_name} split...")
    for i, (img, target) in enumerate(tqdm(dataset)):
        # ChestMNIST a 14 labels (multilabel).
        # Pour votre modèle binaire, on simplifie : 0 = Normal, 1 = Une pathologie détectée.
        label = 1 if np.sum(target) > 0 else 0

        file_name = f"{split_name}_{i}_label_{label}.png"
        file_path = os.path.join(OUTPUT_DIR, file_name)

        # Sauvegarder l'image si elle n'existe pas
        if not os.path.exists(file_path):
            img.save(file_path)

        # Créer l'entrée pour le JSON (BI-Mamba attend frontal et lateral)
        entries.append({
            "image": file_name,
            "frontal": file_path,
            "lateral": file_path,
            "label": label
        })
    return entries

# 3. Générer les listes pour chaque split
data_json = {
    "training": save_and_create_entries(train_dataset, "train"),
    "validation": save_and_create_entries(val_dataset, "val"),
    "test": save_and_create_entries(test_dataset, "test")
}

# 4. Sauvegarder le fichier JSON final
with open(JSON_PATH, "w") as f:
    json.dump(data_json, f, indent=4)

print(f"\n✅ Succès ! {len(data_json['training'])} images réelles prêtes.")
print(f"Le fichier JSON est ici : {JSON_PATH}")
```

***Visualisation des Prédictions (Inférence)***

```python
import torch
import json
import random
import os
import sys
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms
from torchvision.transforms.functional import InterpolationMode

# ==========================================
# 1. CONFIGURATION ET IMPORTS
# ==========================================
project_path = "/content/drive/MyDrive/BI-Mamba-main/mamba-cxr"
if project_path not in sys.path:
    sys.path.append(project_path)

from models_mamba import vim_small_patch16_224_bimambav2_final_pool_mean_abs_pos_embed_with_midclstok_div2 as create_vim_model

CHECKPOINT_PATH = "/content/drive/MyDrive/BI-Mamba-main/mamba-cxr/scripts/OUTPUT/checkpoint.pth"
JSON_PATH = "/content/data/nlst_splits.json"
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# ==========================================
# 2. CHARGEMENT DU MODÈLE
# ==========================================
print(f"🏗️ Construction du modèle...")
model = create_vim_model(num_classes=1)

if os.path.exists(CHECKPOINT_PATH):
    checkpoint = torch.load(CHECKPOINT_PATH, map_location='cpu', weights_only=False)
    state_dict = checkpoint['model'] if 'model' in checkpoint else checkpoint

    # Correction du pos_embed (393 patchs)
    if state_dict['pos_embed'].shape != model.pos_embed.shape:
        model.pos_embed = torch.nn.Parameter(torch.zeros_like(state_dict['pos_embed']))

    model.load_state_dict(state_dict, strict=False)
    model.to(device)
    model.eval()
    print(f"✅ Modèle chargé sur {device}")
else:
    raise FileNotFoundError("Checkpoint introuvable.")

# ==========================================
# 3. TRANSFORMS
# ==========================================
transform = transforms.Compose([
    transforms.Resize(224, interpolation=InterpolationMode.BICUBIC),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))
])

# ==========================================
# 4. INFÉRENCE SUR 10 IMAGES
# ==========================================
with open(JSON_PATH, 'r') as f:
    data = json.load(f)

test_samples = random.sample(data['test'], 10)
plt.figure(figsize=(20, 10))

for i, sample in enumerate(test_samples):
    pil_img = Image.open(sample['frontal']).convert('RGB')
    img_tensor = transform(pil_img).unsqueeze(0).to(device) # (1, 3, 224, 224)

    # --- LA CORRECTION MAGIQUE ---
    # Au lieu de stacker (5D), on concatène sur la LARGEUR (4D)
    # Le modèle va découper 196 patchs sur la gauche et 196 sur la droite.
    # Total : 196 + 196 + 1 (token) = 393. C'est ce qu'attend le pos_embed !
    dual_input = torch.cat([img_tensor, img_tensor], dim=3) # (1, 3, 224, 448)

    with torch.no_grad():
        # On passe l'image large (448px de large)
        output = model(dual_input)
        prob = torch.sigmoid(output).item()
        pred_label = 1 if prob > 0.5 else 0

    # Affichage
    plt.subplot(2, 5, i + 1)
    plt.imshow(pil_img, cmap='gray')
    color = 'green' if sample['label'] == pred_label else 'red'
    plt.title(f"Réel: {sample['label']}\nPrédit: {pred_label}\nProb: {prob:.1%}", color=color)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

***1 - État des lieux***

```python
import matplotlib.pyplot as plt
import re

# Chemin vers ton fichier de sortie
log_path = '/content/vim_small_s16_224.out'

epochs = []
losses = []
aurocs = []

# Extraction des données avec Regex
with open(log_path, 'r') as f:
    content = f.read()

    # Trouve toutes les lignes de Loss d'entraînement
    # Format type: Epoch: [0]  [ 100/468]  loss: 0.6821
    train_matches = re.findall(r"Epoch: \[(\d+)\].*?loss: (\d+\.\d+)", content)

    # Trouve les scores AUROC à la fin de chaque époque
    # Format type: * AUROC 0.7140
    auroc_matches = re.findall(r"\* AUROC (\d+\.\d+)", content)

# Préparation des données pour le graphique
if train_matches:
    # On prend la moyenne de loss par époque pour simplifier le graphique
    for i in range(3): # Pour tes 3 époques
        epoch_losses = [float(m[1]) for m in train_matches if int(m[0]) == i]
        if epoch_losses:
            losses.append(sum(epoch_losses) / len(epoch_losses))
            epochs.append(i + 1)

# Génération des graphiques
plt.figure(figsize=(12, 5))

# Graphique de la LOSS
plt.subplot(1, 2, 1)
plt.plot(epochs, losses, marker='o', color='red', label='Training Loss')
plt.title('Évolution de la Perte (Loss)')
plt.xlabel('Époque')
plt.ylabel('Loss')
plt.grid(True)
plt.legend()

# Graphique de l'AUROC
plt.subplot(1, 2, 2)
if auroc_matches:
    aurocs = [float(a) for a in auroc_matches]
    plt.plot(range(1, len(aurocs)+1), aurocs, marker='s', color='green', label='Test AUROC')
    plt.axhline(y=0.514, color='blue', linestyle='--', label='Baseline (Pre-FT)')
    plt.title('Progression de l\'AUROC')
    plt.xlabel('Époque')
    plt.ylabel('Score AUROC')
    plt.ylim(0.4, 1.0)
    plt.grid(True)
    plt.legend()

plt.tight_layout()
plt.savefig('resultats_bi_mamba.png') # Sauvegarde l'image pour tes slides
plt.show()

print(f"📈 Résumé : AUROC de départ : 0.514 | AUROC final : {aurocs[-1] if aurocs else 'En cours...'}")
```

***Étape 3 : Code de Ré-évaluation et Visualisation***

```python
import torch
import json
import random
import os
import sys
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms
from torchvision.transforms.functional import InterpolationMode

# Configuration des imports
project_path = "/content/drive/MyDrive/BI-Mamba-main/mamba-cxr"
if project_path not in sys.path: sys.path.append(project_path)
from models_mamba import vim_small_patch16_224_bimambav2_final_pool_mean_abs_pos_embed_with_midclstok_div2 as create_vim_model

# 1. Chargement du modèle avec les NOUVEAUX poids
device = 'cuda'
model = create_vim_model(num_classes=1)
checkpoint = torch.load(ckpt_path, map_location='cpu', weights_only=False)
state_dict = checkpoint['model'] if 'model' in checkpoint else checkpoint

# Ajustement automatique du pos_embed (393)
if state_dict['pos_embed'].shape != model.pos_embed.shape:
    model.pos_embed = torch.nn.Parameter(torch.zeros_like(state_dict['pos_embed']))

model.load_state_dict(state_dict, strict=False)
model.to(device).eval()

# 2. Prétraitement
transform = transforms.Compose([
    transforms.Resize(224, interpolation=InterpolationMode.BICUBIC),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))
])

# 3. Visualisation sur les mêmes images (Set de Test)
with open("/content/data/nlst_splits.json", 'r') as f:
    data = json.load(f)

# On reprend 10 images du test
test_samples = random.sample(data['test'], 10)
plt.figure(figsize=(20, 10))

for i, sample in enumerate(test_samples):
    pil_img = Image.open(sample['frontal']).convert('RGB')
    img_tensor = transform(pil_img).unsqueeze(0).to(device)

    # Mode "Panoramique" pour BI-Mamba (224x448)
    dual_input = torch.cat([img_tensor, img_tensor], dim=3)

    with torch.no_grad():
        output = model(dual_input)
        prob = torch.sigmoid(output).item()
        pred_label = 1 if prob > 0.5 else 0

    plt.subplot(2, 5, i + 1)
    plt.imshow(pil_img, cmap='gray')
    color = 'green' if sample['label'] == pred_label else 'red'
    plt.title(f"Réel: {sample['label']}\nPrédit: {pred_label}\nConfiance: {prob:.1%}", color=color, fontweight='bold')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
